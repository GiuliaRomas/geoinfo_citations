# 4. Pré-Processamento

## 4.1. Tratamento das Colunas

```python
df_artigos["titulo"] = df_artigos["titulo"].str.strip()
df_artigos["autores"] = df_artigos["autores"].str.strip()
df_artigos["instituicoes"] = df_artigos["instituicoes"].str.strip()
df_artigos["identificador"] = df_artigos["identificador"].str.strip()
df_artigos["ano"] = pd.to_numeric(df_artigos["ano"], errors="coerce").astype("Int64")
df_artigos["edicao"] = df_artigos["edicao"].str.strip()
```

```python
df_artigos["numero_edicao"] = (pd.to_numeric(df_artigos["edicao"].str.extract(r"(\d+)")[0], errors="coerce").astype("Int64"))
```

```python
df_artigos.rename(columns={"language": "idioma"}, inplace=True)
```

## 4.2. Duplicatas

```python
df_artigos = df_artigos.drop_duplicates(subset="titulo", keep="first").reset_index(drop=True)
```

```python
duplicatas_titulo = df_artigos[df_artigos.duplicated("titulo", keep=False)]
print(f"Títulos duplicados restantes: {len(duplicatas_titulo)}")
```

```python
titulos = df_artigos["titulo"].dropna().unique()

possiveis_duplicatas = []

for i in range(len(titulos)):
    for j in range(i + 1, len(titulos)):
        similaridade = similaridade_titulos(titulos[i], titulos[j])

        if similaridade >= 0.80:
            possiveis_duplicatas.append({
                "titulo_1": titulos[i],
                "titulo_2": titulos[j],
                "similaridade": similaridade
            })
```

```python
pd.DataFrame(possiveis_duplicatas).sort_values("similaridade", ascending=False)
```

```python
titulo = "Improvements to Expectation-Maximization approach for unsupervied classification of remote sensing data"

df_artigos = df_artigos[df_artigos["titulo"] != titulo].copy()
```
